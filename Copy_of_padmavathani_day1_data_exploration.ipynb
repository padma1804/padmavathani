```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Temperature": [30, 32, 28, 35, 31, 29, 34, 33, 27, 36],
    "Humidity": [65, 70, 80, 55, 68, 85, 60, 72, 90, 50],
    "Rainfall": [10, 5, 25, 0, 12, 30, 2, 8, 35, 0],
    "Wind_Speed": [12, 15, 10, 18, 14, 8, 16, 13, 7, 20]
}

df = pd.DataFrame(data)

print(df)
print("Shape:", df.shape)
print(df.isnull().sum())

sns.boxplot(data=df)

plt.title("Weather Data Box Plot")
plt.show()
```
